# Schedule for Section 2

This is the schedule for the semester. This page will be frequently updated
based on how we progress and everything is subject to change. Homeworks and
links to the lecture will be provided here for easy reference to the calendar.
The assignments refer to the homework **following** a class meeting. So if on
a particular date it says watch a video, that is for next time! Recording of
class discussion is always automatically posted after class in NYU Classes.


```python
from IPython.core.display import HTML, Markdown
import itertools

from datetime import datetime, date, timedelta
import calendar

format = "%A, %B %d, %Y"
short_format = "%a %b %d"

# 1. set the first day of classes
first_day = "Wednesday, September 2, 2026"

# 2. set the last day of class
last_day = "Wednesday, December 16, 2026"

# 3. set the recurring days of the week that class meets
days_of_week = ["Wednesday"]

# 4. add in the date of the final exam
final_exam_day = None

# 5. list any exceptions like holidays and the like
#    (a Wednesday listed here becomes a "no class" row automatically)
holidays = {
    "Wednesday, October 14, 2026": "Legislative Day (classes meet on Monday schedule)",
    "Wednesday, November 25, 2026": "Thanksgiving recess",
}
holidays = {datetime.strptime(k, format): v for k, v in holidays.items()}

# 6. add in any additional off-cycle meeting dates
#    (most commonly a legislative day where an off-cycle class meeting happens)
extra_days = {
    # "Tuesday, October 13, 2026": "Legislative Day (classes meet on Wednesday schedule)"
}
extra_days = {datetime.strptime(k, format): v for k, v in extra_days.items()}

# this part runs the script and computes the list of class meeting dates
first_day_f = datetime.strptime(first_day, format)
last_day_f = datetime.strptime(last_day, format)


# generator for all the days between two dates
def daterange(start_date, end_date):
    for n in range(int((end_date - start_date).days + 1)):
        yield start_date + timedelta(n)


class_dates = []
for d in daterange(first_day_f, last_day_f):
    if d in extra_days:
        class_dates.append(d)
    elif calendar.day_name[d.weekday()] in days_of_week:
        class_dates.append(d)

class_dates = sorted(set(class_dates))
```

```python
import re
import os

# ---------------------------------------------------------------------------
# This notebook is a GENERATOR: running it (re)writes `schedule-02.md`, which is
# the file the book actually publishes (see ../_toc.yml). The book build does
# NOT run this notebook -- so after editing dates/content above, re-run this
# notebook and commit the regenerated schedule-02.md.
#
# You can also hand-edit schedule-02.md directly; just know a re-run overwrites
# those manual edits, so prefer editing the content list above when possible.
# ---------------------------------------------------------------------------

OUTPUT_FILE = "schedule-02.md"
SECTION_TITLE = "# Schedule for Section 2"

PREAMBLE = """This is the schedule for the semester. This page will be frequently updated
based on how we progress and everything is subject to change. Homeworks and
links to the lecture will be provided here for easy reference to the calendar.
The assignments refer to the homework **following** a class meeting. So if on
a particular date it says watch a video, that is for next time!"""


def clean_markup(text):
    """Convert the small bits of inline HTML used in the content dicts to
    idiomatic markdown: <b>..</b> -> **..**, and drop stray <br> tags."""
    text = re.sub(r"</?b>", "**", text)
    text = re.sub(r"<br\s*/?>", " ", text)
    return text


def md_links_to_doc_roles(text):
    """Convert [label](../path/to/page.html) -> {doc}`label <../path/to/page>`.

    Jupyter Book cross-references pages with the {doc} role and a path WITHOUT
    the .html extension, so we strip .html and rewrite the markdown link.
    External http(s) links are left as-is.
    """
    def repl(m):
        label, url = m.group(1), m.group(2)
        if url.startswith("http"):
            return f"[{label}]({url})"
        url = re.sub(r"\.html($|#)", r"\1", url)  # drop .html (keep any #anchor)
        return f"{{doc}}`{label} <{url}>`"

    return re.sub(r"\[([^\]]+)\]\(([^)]+)\)", repl, text)


def render(text):
    return md_links_to_doc_roles(clean_markup(text))


def as_bullets(value, indent="  "):
    """Render a string or list/tuple of strings as markdown bullet lines."""
    items = value if isinstance(value, (list, tuple)) else [value]
    items = [i for i in items if str(i).strip()]
    return "\n".join(f"{indent}- {render(i)}" for i in items)


# ---- build the markdown body, one section (###) per date -------------------
sections = []
for d in class_dates_info:
    info = class_dates_info[d]
    heading = f"### {datetime.strftime(d, '%a, %b %d')}"
    if info.get("NoClass"):
        # for a true holiday the reason lives in the Agenda; surface it here.
        reason = info.get("Agenda", "")
        reason = re.sub(r"</?b>|No class\s*-\s*", "", reason).strip() if isinstance(reason, str) else ""
        heading += f"  (No class{' — ' + reason if reason else ''})"
    if info.get("Warning"):
        heading += "  ⚠️"

    block = [heading]

    # skip the Agenda bullet for holidays -- the reason is already in the heading
    if not info.get("NoClass"):
        agenda_bullets = as_bullets(info.get("Agenda", "TBD"))
        if agenda_bullets:
            block.append("- Agenda")
            block.append(agenda_bullets)

    assignment_bullets = as_bullets(info.get("Assignments", ""))
    if assignment_bullets:
        block.append("- Assignments")
        block.append(assignment_bullets)

    sections.append("\n".join(block))

schedule_md = "\n\n".join([SECTION_TITLE, PREAMBLE, *sections]) + "\n"

# ---- write it out next to this notebook ------------------------------------
out_path = os.path.join(os.getcwd(), OUTPUT_FILE)
with open(out_path, "w") as f:
    f.write(schedule_md)
print(f"Wrote {len(sections)} dated sections to {out_path}")
```
